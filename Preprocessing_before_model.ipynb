```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Charger le fichier CSV
file_path = 'data\custom_data\MSATTS_updated.csv'
df = pd.read_csv(file_path)


# Diviser le dataset en train et validation
train_df, val_df = train_test_split(df, train_size=0.8, shuffle=True, random_state=42)
# Chemins de sortie pour les fichiers texte
train_file = 'data\custom_data/train.txt'
val_file = 'data\custom_data/validation.txt'

# Fonction pour écrire les données dans un fichier texte
def write_to_file(df, output_file):
    with open(output_file, 'w', encoding='utf-8') as f:
        for index, row in df.iterrows():
            # Format : "Nom du fichier|Texte diacrité|Speaker_ID"
            line = f"{row['Nom du fichier']}|{row['Texte diacrité']}|{row['Speaker_ID']}\n"
            f.write(line)

# Écrire les fichiers train.txt et validation.txt
write_to_file(train_df, train_file)
write_to_file(val_df, val_file)

# Confirmer la création des fichiers
print(f"Fichier d'entraînement créé : {train_file} ({len(train_df)} lignes)")
print(f"Fichier de validation créé : {val_file} ({len(val_df)} lignes)")
```

```python
# Opening training and validation datasets

with open('data\custom_data/train.txt', 'r', encoding='utf-8') as file: # Changed encoding to 'utf-8'
  train_data = file.readlines()

with open('data\custom_data/validation.txt', 'r', encoding='utf-8') as file: # Changed encoding to 'utf-8'
    val_data = file.readlines()
train_data = [line.strip().split('|') for line in train_data]
val_data = [line.strip().split('|') for line in val_data]
```

```python
import pandas as pd


# Create a data frame from the content of the .txt file
train_data = pd.DataFrame(train_data, columns=['Nom du fichier', 'Texte diacrité','Speaker_ID'])
val_data = pd.DataFrame(val_data, columns=['Nom du fichier', 'Texte diacrité','Speaker_ID'])

# Display the data frame
train_data.head(10)
```

```python
import text


#Helper function that turns arabic text into phonemes
def preprocess(txt):
  t_phon = text.arabic_to_buckwalter(txt)
  t_phon = text.buckwalter_to_phonemes(t_phon)
  return t_phon
```

```python
train_phonemes,val_phonemes = [], []

for _, row in train_data.iterrows():
  row['Nom du fichier'] += ''
  train_phonemes.append(preprocess(row['Texte diacrité']))

for _,row in val_data.iterrows():
  row['Nom du fichier'] += ''
  val_phonemes.append(preprocess(row['Texte diacrité']))
train_data = train_data.assign(Phonemes=train_phonemes)
val_data = val_data.assign(Phonemes=val_phonemes)
train_data = train_data.drop('Texte diacrité', axis=1)
val_data = val_data.drop('Texte diacrité', axis=1)
```

```python

import csv
train_data.to_csv('data\custom_data/ready_train.txt', sep=' ', header=None, index=None, quoting=csv.QUOTE_NONNUMERIC)
val_data.to_csv('data\custom_data/ready_val.txt', sep=' ', header=None, index=None, quoting=csv.QUOTE_NONNUMERIC)
```

```python
# %%
import os
import torch
import librosa
import numpy as np
import torch.nn.functional as F
from tqdm import tqdm
import pandas as pd
from utils.audio import MelSpectrogram
from utils import write_lines_to_file

# %% CONFIG

csv_path = 'data\custom_data\MSATTS_updated.csv'
pitch_dict_path = 'data\custom_data/pitch_dict.pt'

# Load the dataset to determine which files to process
df = pd.read_csv(csv_path)
waves = df['Nom du fichier'].tolist()  # Get full paths from the CSV
print(f"{len(waves)} wave files found in {csv_path}")

mel_trf = MelSpectrogram()

# %% Load existing pitch dictionary if it exists
pitch_dict = {}
if os.path.exists(pitch_dict_path):
    try:
        pitch_dict = torch.load(pitch_dict_path)
        print(f"Loaded existing pitch dictionary with {len(pitch_dict)} entries.")
    except Exception as e:
        print(f"Error loading pitch dictionary: {e}. Starting with an empty dictionary.")
else:
    print("No existing pitch dictionary found. Starting fresh.")

# %% Extract pitch (f0) values

for i, wav_path in tqdm(enumerate(waves), total=len(waves)):
    wav_name = os.path.basename(wav_path)  # Extract filename for dictionary key

    # Skip if the file is already processed
    if wav_name in pitch_dict:
        continue

    try:
        # Check if file exists
        if not os.path.exists(wav_path):
            print(f"Warning: File not found: {wav_path}. Skipping.")
            continue

        # Load and process the audio
        wav, sr = librosa.load(wav_path, sr=mel_trf.sample_rate)
        mel_spec = mel_trf(torch.tensor(wav)[None])[0]  # [mel_bands, T]

        # Estimate pitch
        pitch_mel, voiced_flag, voiced_probs = librosa.pyin(
            wav, sr=mel_trf.sample_rate,
            fmin=librosa.note_to_hz('C2'),
            fmax=librosa.note_to_hz('C7'),
            frame_length=mel_trf.win_length,
            hop_length=mel_trf.hop_length)

        pitch_mel = np.where(np.isnan(pitch_mel), 0., pitch_mel)  # Set NaN to zero
        pitch_mel = torch.from_numpy(pitch_mel)
        pitch_mel = F.pad(pitch_mel, (0, mel_spec.size(1) - pitch_mel.size(0)))  # Pad to mel length

        # Store in dictionary using filename as key
        pitch_dict[wav_name] = pitch_mel

        # Save intermediate dict every 10 files
        if i % 10 == 0 and i > 0:
            torch.save(pitch_dict, pitch_dict_path)
            print(f"Intermediate save at {i} files: {len(pitch_dict)} entries.")

    except Exception as e:
        print(f"Error processing {wav_name}: {e}")
        continue

# Final save
torch.save(pitch_dict, pitch_dict_path)
print(f"Final pitch dictionary saved with {len(pitch_dict)} entries.")
```

```python
# %% Calculate pitch mean and std

rmean = 0
rvar = 0
ndata = 0

for pitch_mel in pitch_dict.values():
    pitch_mel = np.where(np.isnan(pitch_mel), 0.0, pitch_mel)

    pitch_mel_ = pitch_mel[pitch_mel > 1]
    if len(pitch_mel_) == 0:  # Skip if no valid pitch values
        continue

    p_mean = np.mean(pitch_mel_)
    p_var = np.var(pitch_mel_)
    p_len = len(pitch_mel_)

    rvar = ((ndata-1)*rvar + (p_len-1)*p_var) / (ndata + p_len - 1) + \
            ndata*p_len*(p_mean - rmean)**2 / ((ndata + p_len)*(ndata + p_len - 1))

    rmean = (p_len*p_mean + ndata*rmean) / (p_len + ndata)

    ndata += p_len

mean, std = rmean, np.sqrt(rvar)
print('Mean:', mean)
print('Std:', std)

write_lines_to_file(path='/content/drive/MyDrive/Stage_pfe/models_tts/mean_std.txt',
                    lines=[f"mean: {mean}",
                           f"std: {std}"])
```
